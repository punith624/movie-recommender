```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from src.data_loader import load_movielens, create_mappings, train_test_split_ratings
from src.matrix_factorization import MatrixFactorization
from src.recommend import get_top_n_recommendations

# Load MovieLens ratings and movies
ratings, movies = load_movielens()
ratings.shape, movies.shape
```

```python
print('Users:', ratings['userId'].nunique())
print('Movies:', ratings['movieId'].nunique())

n_users = ratings['userId'].nunique()
n_movies = ratings['movieId'].nunique()

sparsity = 1 - (len(ratings) / (n_users * n_movies))
print('Matrix sparsity:', sparsity)

ratings['rating'].hist(bins=10)
plt.title("Rating Distribution")
plt.show()
```

```python
# Split ratings into train/test
train_df, test_df = train_test_split_ratings(ratings, test_size=0.2)

# Map userId/movieId → matrix index
user2idx, idx2user, movie2idx, idx2movie = create_mappings(ratings)

def to_triplets(df):
    trip = []
    for r in df.itertuples():
        if r.userId in user2idx and r.movieId in movie2idx:
            trip.append((user2idx[r.userId], movie2idx[r.movieId], r.rating))
    return trip

train_triplets = to_triplets(train_df)
test_triplets = to_triplets(test_df)

len(train_triplets), len(test_triplets)
```

```python
mf = MatrixFactorization(
    n_users=len(user2idx),
    n_items=len(movie2idx),
    k=30,          # latent factors
    lr=0.007,      # learning rate
    reg=0.02,      # regularization
    n_epochs=30,   # training epochs
    verbose=True
)

history = mf.fit(train_triplets, val_data=test_triplets)
```

```python
plt.plot(history['train_rmse'], label='Train')
plt.plot(history['val_rmse'], label='Validation')
plt.xlabel("Epoch")
plt.ylabel("RMSE")
plt.legend()
plt.title("Matrix Factorization Training Curve")
plt.show()
```

```python
# Save trained model
mf.save("models/mf_model")

# Build train mask: used to avoid recommending movies the user has seen
train_mask = np.zeros((len(user2idx), len(movie2idx)), dtype=bool)
for (u, i, _) in train_triplets:
    train_mask[u, i] = True

# Recommend for user index 0
example_user_idx = 0
recs = get_top_n_recommendations(
    mf, example_user_idx, idx2movie, movies, train_mask=train_mask, n=10
)

recs[['title', 'pred_score']]
```
